#Dashboard de desempeño comercial 2024–2025
##Empresa: **Andes Retail Group**


## 🧭 Introducción

**Andes Retail Group** es una empresa de retail con operaciones en Perú, Chile y Colombia que comercializa productos en cuatro categorías:
- 🖥️ Electrónica
- 👕 Ropa
- ⚽ Deportes
- 🏠 Hogar

La dirección ejecutiva necesita un dashboard interactivo que permita entender el desempeño comercial de los años 2024–2025. Actualmente, la información está dispersa en datos transaccionales y no existe una visión clara que permita responder preguntas estratégicas sobre ventas, rentabilidad y comportamiento de clientes.

👉 La misión: transformar estos datos en información visual, clara y accionable.

### 💡 Preguntas del negocio
- ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
- ¿Qué segmentos de clientes aportan mayor ingreso y rentabilidad?
- ¿Qué categorías de producto tienen mayor impacto en el negocio?
- ¿Existen diferencias relevantes entre países o regiones?
- ¿Qué patrones temporales se observan a lo largo del año?
- ¿Dónde podrían existir oportunidades de mejora comercial?

## 🔹 3. Diseño y planificación del Dashboard (antes y durante la construcción)


### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?

####**KPI's empleados:**

- **Ingresos Totales:** Es la tarjeta principal porque responde directamente "cuánto vendimos". La consideramos como base de todo el análisis.
- **Crecimiento de Ingresos 2024 vs 2025:** Este KPI muestra si el negocio creció o decreció en términos porcentuales.
- **Margen de Rentabilidad %:** Un crecimiento en ingresos no siempre significa salud financiera — la dirección necesita saber si ese crecimiento viene acompañado de rentabilidad o si estamos vendiendo más pero ganando menos por unidad.
- **Ticket Promedio (Ingreso / # Pedidos):** Este KPI ayuda a diagnosticar por qué cambió el ingreso. Es un primer indicio de comportamiento de compra, que conecta con la pregunta de segmentos más adelante.

####**Gráficos empleados:**

- **Gráfico de tiempo:** Línea — Ingreso mensual 2024 vs 2025
  Responde: ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
Se eligió gráfico de líneas porque la pregunta es explícitamente sobre evolución y tendencia. De manera que la línea comunica dirección de forma más inmediata. Superponer ambos años en dos líneas de distinto color permite comparar estacionalidad y crecimiento mes a mes sin necesitar cambiar de filtro.

- **Gráfico comparación geográfica:** Barras horizontales — Ingreso Total por País
  Responde: ¿Existen diferencias relevantes entre países?
Se eligió barras horizontales sobre mapa geográfico por dos razones: el número reducido de países (Perú, Chile, Colombia) no justifica un mapa que generaría demasiado espacio vacío, y las limitaciones del visual de mapa proporcional en Power BI Desktop versión gratuita. Las barras permiten lectura inmediata del ranking por ingreso.

- **Gráfico comparación productos:** Barras verticales — Ingreso Total por Categoría
  Responde: ¿Qué categorías de producto tienen mayor impacto en el negocio?
Con solo 4 categorías las barras verticales son más legibles que un treemap o dona, y permiten comparar valores exactos entre categorías. Al interactuar con el slicer de año revela que Deportes lidera en 2024 pero Hogar toma el liderazgo en 2025.

- **Gráfico comparación clientes:** Treemap — Ingreso Total por Segmento_Cliente
  Responde: ¿Qué segmentos de clientes aportan mayor ingreso?
El treemap muestra proporción y jerarquía de un vistazo. Se complementa con Ticket Promedio y Total de pedidos como tooltips para distinguir si el liderazgo de Premium responde a volumen de pedidos o a mayor gasto por transacción.


####**Descripción de la jerarquía visual**.

**Portada**. Página de entrada con imagen de contexto retail, título del proyecto y subtítulo. Genera una primera impresión profesional antes de entrar al análisis.

**Título de página**. "Overview Ejecutivo" visible en la esquina superior izquierda, presente como orientación al usuario.

**Fila superior**. 4 tarjetas KPI ancladas arriba. Son el primer elemento que lee la dirección ejecutiva porque resumen el estado del negocio en números clave antes de explorar los gráficos.

**Centro, mayor tamaño**. Gráfico de líneas temporal como protagonista visual de la página, ocupa el ancho completo porque responde la pregunta central del Overview.

**Fila inferior dividida en 3 columnas**. Barras por país, barras por categoría y treemap de segmentos. Información secundaria que profundiza después de entender la tendencia general.

**Jerarquía de color**. La tarjeta de Crecimiento % usa color de alerta (rosa/salmón) para dirigir la atención al dato negativo (-6.80%) de forma inmediata.


**Filtro(s)**

**Año**. Slicer con opciones 2024 y 2025. Permite aislar o comparar ambos períodos de forma interactiva. Se posiciona en la esquina superior derecha fuera del flujo de lectura principal para no competir visualmente con los KPIs. Al activarlo afecta todos los gráficos de la página simultáneamente.

### Vista detalle - ¿Por qué se han caido/subido/mantenido los ingresos?


####**Gráficos empleados:**

- **Gráfico de rentabilidad por país:** Pastel — Margen % por País
  Responde: ¿Existen diferencias relevantes entre países en términos de rentabilidad?
Se eligió gráfico de pastel porque los tres países muestran valores muy similares (~34% cada uno). El pastel comunica exactamente eso: equilibrio proporcional, siendo que las barras darían una impresión errónea de disparidad.

- **Gráfico de rentabilidad por categoría:** Barras verticales — Margen % por Categoría
  Responde: ¿Qué categorías de producto tienen mayor rentabilidad?
Se mantiene en barras, ahora verticales, para consistencia con el Overview, pero aquí el eje muestra Margen % en vez de Ingreso Total. Permite contrastar directamente con el gráfico de ingresos de la página anterior — Electrónica lidera en rentabilidad aunque no en volumen de ventas.

- **Gráfico de rentabilidad por segmento:** Pastel — Margen % por Segmento_Cliente
  Responde: ¿Qué segmentos de clientes aportan mayor rentabilidad?
Igual que en el gráfico por país, el pastel es la elección correcta porque los tres segmentos rondan el mismo valor (~34%). Revela que Premium, a pesar de liderar en ingreso y ticket promedio en el Overview, no tiene ventaja competitiva en rentabilidad.

- **Gráfico de patrones temporales:** Barras verticales — Margen % por Estación
  Responde: ¿Qué patrones temporales se observan a lo largo del año?
Se usa la columna Estación para analizar si la rentabilidad varía según época del año. El gráfico confirma que el margen se mantiene estable (~33-35%) en todas las estaciones, aunque Verano lidera en ingresos y Primavera en rentabilidad.

- **Tabla detalle:** Tabla — País, Categoría, Segmento, Nivel_Venta, Ticket Promedio, Margen %, Ingreso Total
  Responde: ¿Dónde podrían existir oportunidades de mejora comercial?
La tabla permite a la dirección ejecutiva hacer drill manual cruzando las tres dimensiones principales (país, categoría, segmento) con las métricas clave.


####**Descripción de la jerarquía visual**.

**Título de página**. "Vista Detalle" en la esquina superior izquierda, mantiene consistencia con la página anterior.

**Fila superior**. El gráfico de pastel por país se ubica al inicio de la página porque al hacer click en cualquier país, todos los demás gráficos se actualizan automáticamente mostrando solo la información de ese país. Esto permite a la dirección explorar la rentabilidad por región sin necesidad de un filtro adicional. La tabla detalle en paralelo complementa esta interacción mostrando el desglose granular en tiempo real.

**Fila inferior**. Dividida en 3 columnas: barras de estaciones, pastel de segmentos y barras de categoría. Los tres responden dimensiones distintas del análisis de rentabilidad de forma compacta.

**Jerarquía de color**. Se mantiene la paleta del Overview para consistencia visual entre páginas. Cada visual está contenido en un card blanco con bordes redondeados que mejora la legibilidad y separa claramente cada análisis.


**Filtros**.

**Nivel de Venta**. Slicer con opciones Venta Alta y Venta Baja, basado en la columna calculada Nivel_venta creada en Power Query. Permite analizar si los patrones de rentabilidad, categoría y país se mantienen o cambian al filtrar por nivel de venta. Al activarlo afecta todos los gráficos de la página simultáneamente.

**Año**. Slicer con opciones 2024 y 2025, consistente con el Overview. Permite comparar si los patrones de rentabilidad cambiaron entre ambos períodos.

### Narrativa del Dashboard (Modelo SCQA)


#### 🖥️ Vista General (Overview)

**S (Situación):**  
Andes Retail Group opera en tres países (Perú, Chile y Colombia) comercializando cuatro categorías de producto. La dirección ejecutiva necesita una visión consolidada del desempeño comercial 2024–2025 para tomar decisiones estratégicas.

**C (Complicación):**  
La información estaba dispersa en datos transaccionales sin una visión integrada. Al consolidarla, se detecta que los ingresos totales cayeron un -6.80% entre 2024 y 2025, con cambios en el liderazgo de categorías. Deportes dominaba en 2024 pero Hogar tomó el primer lugar en 2025.

**Q (Pregunta):**   
¿Cómo ha evolucionado el desempeño comercial entre 2024 y 2025, y qué dimensiones del negocio explican ese cambio?

**A (Respuesta):**  
El ingreso total de **5,534** millones refleja una caída generalizada más pronunciada en los primeros meses de 2025, con un patrón estacional consistente en ambos años (pico en diciembre, valle en junio-julio). Perú lidera geográficamente de forma consistente. El segmento Premium genera el mayor valor por transacción (**1,700** de ticket promedio) aunque el segmento Estándar sostiene la operación en volumen de pedidos.

---

#### 🔎 Vista Detalle

**S (Situación):**  
El Overview revela una caída en ingresos del -6.80%, con diferencias entre categorías y segmentos en términos de volumen de ventas. El siguiente paso natural es analizar si esas diferencias se replican en rentabilidad.

**C (Complicación):**  
Al analizar el Margen %, se descubre una desconexión entre las categorías que generan mayor ingreso y las más rentables. Electrónica y Ropa lideran en margen aunque no en ventas. Más sorprendente aún: el margen se mantiene en ~34% sin importar el país, segmento o estación del año, lo que sugiere que no hay palancas de eficiencia siendo aprovechadas en ningún segmento específico.

**Q (Pregunta):**  
¿Dónde están las oportunidades de mejora comercial si la rentabilidad es uniforme en toda la operación?

**A (Respuesta):**  
Las oportunidades no están en eficiencia de costos (el margen es consistente) sino en mix de producto y segmento. La empresa podría potenciar Electrónica y Ropa, que son más rentables, sobre Deportes y Hogar. El segmento Estándar, que lidera en volumen pero tiene ticket promedio de **970**, representa una oportunidad de upselling. El segmento Económico (**508** de ticket) es el de mayor potencial de crecimiento por transacción.

**Un hallazgo crítico emerge del análisis:** El segmento Premium genera tres veces más ingreso por transacción que el Económico (**1,700 vs 508**), sin embargo los tres segmentos operan con el mismo margen de rentabilidad (~34%). Esto invita a la dirección a cuestionar si los costos asociados a retener y atender clientes Premium están justificados frente a los resultados en rentabilidad, o si los recursos deberían redistribuirse hacia los segmentos de mayor volumen.
